# Quadratic programming

Minimize $(1/2)x^T P x + q^T x$ subject to $G x \leq h$ and $A x = b$.

```python
import cvxpy as cp
import numpy as np

m, n, p = 15, 10, 5
np.random.seed(1)
P = np.random.randn(n, n)
P = P.T @ P
q = np.random.randn(n)
G = np.random.randn(m, n)
h = G @ np.random.randn(n)
A = np.random.randn(p, n)
b = np.random.randn(p)

x = cp.Variable(n)
prob = cp.Problem(cp.Minimize((1/2)*cp.quad_form(x, P) + q.T @ x), [G@x <= h, A@x == b])
prob.solve()
print('Optimal value', prob.value)
print('x', x.value)
```
