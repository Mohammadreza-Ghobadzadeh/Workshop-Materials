# Optimal Control

Linear dynamics $x_{t+1} = A x_t + B u_t$. Minimize sum of stage costs with input bounds and a terminal constraint $x_T = 0$.

```python
import numpy as np
import cvxpy as cp
import matplotlib.pyplot as plt

np.random.seed(1)
n, m, T = 8, 2, 50
alpha, beta = 0.2, 3
A = np.eye(n) - alpha * np.random.rand(n, n)
B = np.random.randn(n, m)
x0 = beta * np.random.randn(n)

x = cp.Variable((n, T + 1))
u = cp.Variable((m, T))
cost = 0
constr = [x[:, 0] == x0, x[:, T] == 0]
for t in range(T):
    cost += cp.sum_squares(x[:, t + 1]) + cp.sum_squares(u[:, t])
    constr += [x[:, t + 1] == A @ x[:, t] + B @ u[:, t], cp.norm(u[:, t], 'inf') <= 1]
prob = cp.Problem(cp.Minimize(cost), constr)
prob.solve()
print(prob.status, prob.value)

plt.plot(u.value.T)
plt.title('Optimal inputs')
plt.xlabel('t')
plt.show()
```
