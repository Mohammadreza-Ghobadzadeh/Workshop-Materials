# Supply Chain Problem

How many trucks should we send from depots to warehouses and from warehouses to restaurants to meet quinoa demand while minimizing kilometers traveled? A truck carries 10 boxes.

```python
import cvxpy as cp
import numpy as np

Routes = {
    'D1_W1': 5, 'D1_W2': 3, 'D1_W3': 4,
    'D2_W1': 2.2, 'D2_W2': 4, 'D2_W3': 2,
    'W1_R1': 1, 'W1_R2': 1.2, 'W1_R3': 1.4, 'W1_R4': 1.5, 'W1_R5': 1.7,
    'W2_R1': 1.2, 'W2_R2': 1.4, 'W2_R3': 1.6, 'W2_R4': 1.5, 'W2_R5': 1.1,
    'W3_R1': 1.4, 'W3_R2': 1.6, 'W3_R3': 1.8, 'W3_R4': 1.7, 'W3_R5': 0.9,
}
costs = np.array([float(Routes[c]) for c in Routes])
x = cp.Variable(shape=len(Routes), integer=True)
objective = cp.Minimize(x.T @ costs)

def idx(route):
    return list(Routes.keys()).index(route)

constraints = [
    x[idx('W1_R1')] + x[idx('W1_R2')] + x[idx('W1_R3')] + x[idx('W1_R4')] + x[idx('W1_R5')] <= 500,
    x[idx('W2_R1')] + x[idx('W2_R2')] + x[idx('W2_R3')] + x[idx('W2_R4')] + x[idx('W2_R5')] <= 1200,
    x[idx('W3_R1')] + x[idx('W3_R2')] + x[idx('W3_R3')] + x[idx('W3_R4')] + x[idx('W3_R5')] <= 800,
    x[idx('W1_R1')] + x[idx('W2_R1')] + x[idx('W3_R1')] >= 300,
    x[idx('W1_R2')] + x[idx('W2_R2')] + x[idx('W3_R2')] >= 350,
    x[idx('W1_R3')] + x[idx('W2_R3')] + x[idx('W3_R3')] >= 450,
    x[idx('W1_R4')] + x[idx('W2_R4')] + x[idx('W3_R4')] >= 400,
    x[idx('W1_R5')] + x[idx('W2_R5')] + x[idx('W3_R5')] >= 550,
    x[idx('D1_W1')] + x[idx('D2_W1')] >= x[idx('W1_R1')] + x[idx('W1_R2')] + x[idx('W1_R3')] + x[idx('W1_R4')] + x[idx('W1_R5')],
    x[idx('D1_W2')] + x[idx('D2_W2')] >= x[idx('W2_R1')] + x[idx('W2_R2')] + x[idx('W2_R3')] + x[idx('W2_R4')] + x[idx('W2_R5')],
    x[idx('D1_W3')] + x[idx('D2_W3')] >= x[idx('W3_R1')] + x[idx('W3_R2')] + x[idx('W3_R3')] + x[idx('W3_R4')] + x[idx('W3_R5')],
    x >= 0,
]
problem = cp.Problem(objective, constraints)
print('Optimal:', problem.solve())
print(x.value)
for i, name in enumerate(Routes):
    if x.value[i] > 0:
        print('Route', name, x.value[i])
```
