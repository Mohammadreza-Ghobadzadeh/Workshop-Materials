# Constraint
CVXPY has seven types of constraints: non-positive, equality or zero, positive semidefinite, second-order cone, exponential cone, 3-dimensional power cones, and N-dimensional power cones

```python
import cvxpy as cp

x = cp.Variable()
y = cp.Variable()

constraints = [x + y == 1,
               x - y >= 1]
obj = cp.Minimize((x - y)**2)
prob = cp.Problem(obj, constraints)
prob.solve()
print("status:", prob.status)
print("optimal value", prob.value)
print("optimal var", x.value, y.value)
print('=====================================')

cons1=cp.constraints.Zero(x+y-1)
cons2=cp.constraints.NonPos(y-x+1)
constraints=[cons1,cons2]
obj = cp.Minimize((x - y)**2)
prob = cp.Problem(obj, constraints)
prob.solve()
print("status:", prob.status)
print("optimal value", prob.value)
print("optimal var", x.value, y.value)
```

# Problem

1- Define objective function (Minimize or Maximize)\
2- Define constraints\
3- Build the problem {class cvxpy.Problem (objective: Minimize | Maximize, constraints: list [Constraint])}\
4- Solve the problem using solver\
5- Use the .value command to see the solution of the optimization problem\
6- Use the .status command to see the solution status

```python
# An example
# Minimizing x**2 constrained to x<=2
import cvxpy as cp
x = cp.Variable()
constraints = [x<=2]
obj = cp.Minimize((x)**2)
prob = cp.Problem(obj, constraints)
prob.solve()
print("status:", prob.status)
print("optimal value", prob.value)
print("optimal var", x.value)
```

**Solve method options**\
solve(solver=None, verbose=False, gp=False, qcp=False, requires_grad=False, enforce_dpp=False, ignore_dpp=False, **kwargs)

**Problem statuses**
* OPTIMAL
* INFEASIBLE
* UNBOUNDED
* OPTIMAL_INACCURATE
* INFEASIBLE_INACCURATE
* UNBOUNDED_INACCURATE
* INFEASIBLE_OR_UNBOUNDED

```python
import cvxpy as cp
x = cp.Variable()
prob = cp.Problem(cp.Minimize(x), [x >= 1, x <= 0])
prob.solve()
print("status:", prob.status)
print("optimal value", prob.value)

prob = cp.Problem(cp.Minimize(x))
prob.solve()
print("status:", prob.status)
print("optimal value", prob.value)
```

*If the solver completely fails to solve the problem, CVXPY throws a SolverError exception

**Problems are immutable, meaning they cannot be changed after they are created. To change the objective or constraints, create a new problem

```python
import cvxpy as cp
x = cp.Variable()
constraints = [x<=2]
obj = cp.Minimize((x)**2)
prob = cp.Problem(obj, constraints)
prob2 = cp.Problem(cp.Minimize((x-1)**2),prob.constraints)
prob2.solve()
print("status:", prob2.status)
print("optimal value", prob2.value)
print("optimal var", x.value)
print('=================================')
prob3 = cp.Problem(prob.objective,[x<=-1,x>=-2])
prob3.solve()
print("status:", prob3.status)
print("optimal value", prob3.value)
print("optimal var", x.value)
```
