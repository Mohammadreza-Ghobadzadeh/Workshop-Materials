# CVXPY

CVXPY is a Python-embedded modeling language for convex optimization. Documentation: https://www.cvxpy.org/

```python
import cvxpy as cp
import numpy as np

x = cp.Variable()
a = cp.Parameter()
b = cp.Parameter(value=2)

# shapes
scalar = cp.Variable()
vec = cp.Variable(5)
col = cp.Variable((5, 1))
mat = cp.Variable((4, 7))
Xpsd = cp.Variable((10, 10), PSD=True)
```

## A first problem

Minimize $x^2 + y^2$ subject to $x + y >= 1$, $x >= 0$, $y >= 0$.

```python
x = cp.Variable()
y = cp.Variable()
prob = cp.Problem(cp.Minimize(x**2 + y**2), [x + y >= 1, x >= 0, y >= 0])
prob.solve()
print(prob.status, prob.value, x.value, y.value)
```
