# Maximizing the volume of a box

Adapted from Boyd, Kim, Vandenberghe, and Hassibi, A Tutorial on Geometric Programming.

Maximize $hwd$ subject to wall and floor area limits and aspect-ratio bounds.

```python
import cvxpy as cp

A_wall, A_flr = 100, 10
alpha, beta, gamma, delta = 0.5, 2, 0.5, 2
h = cp.Variable(pos=True, name='h')
w = cp.Variable(pos=True, name='w')
d = cp.Variable(pos=True, name='d')
volume = h * w * d
constraints = [2*(h*w + h*d) <= A_wall, w*d <= A_flr, h/w >= alpha, h/w <= beta, d/w >= gamma, d/w <= delta]
problem = cp.Problem(cp.Maximize(volume), constraints)
print(problem.is_dcp(), problem.is_dgp())
problem.solve(gp=True)
print(problem.value, h.value, w.value, d.value)
```
